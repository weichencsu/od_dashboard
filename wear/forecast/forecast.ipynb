```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# 原始数据点
x = np.array([0, 0.595, 0.668])
y = np.array([105, 96.05, 92.068])

# 线性拟合
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

# 计算x=0.346和x=0.738时的y值
x_target1 = 0.346
y_target1 = slope * x_target1 + intercept

x_target2 = 0.738
y_target2 = slope * x_target2 + intercept

print(f"拟合参数:")
print(f"斜率 (slope): {slope:.4f}")
print(f"截距 (intercept): {intercept:.4f}")
print(f"相关系数 R: {r_value:.4f}")
print(f"R²: {r_value**2:.4f}")

print(f"\n当x = {x_target1}时，y = {y_target1:.4f}")
print(f"当x = {x_target2}时，y = {y_target2:.4f}")

# 生成拟合直线的数据点
x_fit = np.linspace(0, 0.8, 100)
y_fit = slope * x_fit + intercept

# 绘制图表
plt.figure(figsize=(12, 7))
plt.scatter(x, y, color='red', s=100, label='原始数据点', zorder=5)
plt.plot(x_fit, y_fit, color='blue', linewidth=2, label=f'线性拟合: y = {slope:.2f}x + {intercept:.2f}', zorder=3)

# 标注原始数据点
for i in range(len(x)):
    plt.annotate(f'({x[i]}, {y[i]})', (x[i], y[i]), xytext=(5, 5), 
                textcoords='offset points', fontsize=9)

# 标注两个预测点
plt.scatter(x_target1, y_target1, color='green', s=150, marker='^', 
           label=f'预测点1 ({x_target1}, {y_target1:.2f})', zorder=6)
plt.annotate(f'({x_target1}, {y_target1:.2f})', (x_target1, y_target1), xytext=(5, 5), 
            textcoords='offset points', fontsize=9, color='green')

plt.scatter(x_target2, y_target2, color='purple', s=150, marker='s', 
           label=f'预测点2 ({x_target2}, {y_target2:.2f})', zorder=6)
plt.annotate(f'({x_target2}, {y_target2:.2f})', (x_target2, y_target2), xytext=(5, 5), 
            textcoords='offset points', fontsize=9, color='purple')

plt.xlabel('x')
plt.ylabel('y')
plt.title('线性拟合结果 - 包含两个预测点')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-0.05, 0.8)
plt.ylim(80, 110)

# 添加R²信息到图中
plt.text(0.02, 0.98, f'R² = {r_value**2:.4f}', transform=plt.gca().transAxes, 
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

plt.tight_layout()
plt.show()

print(f"\n拟合方程: y = {slope:.4f}x + {intercept:.4f}")
print(f"注意事项: x=0.738超出了原始数据范围[0, 0.677]，这是外推预测")
```
